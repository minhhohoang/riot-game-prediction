```python
import requests
import pandas as pd
import time
import json
from lolalytics_api import matchup
```

```python
PATCH = "16.19.1"
LOLALYTICS_PATCH = "16.19"

LANE = "top"
RANK = "emerald"


url = (
    f"https://ddragon.leagueoflegends.com/"
    f"cdn/{PATCH}/data/en_US/champion.json"
)

response = requests.get(url)
data = response.json()

champions = [
    champ["id"].lower()
    for champ in data["data"].values()
]

print(f"Found {len(champions)} champions")
print(champions[:10])
```

```python
import json
import pandas as pd
import time

test_champions = [
    "riven",
    "malphite",
    "garen",
    "fiora",
    "camille"
]

rows = []

for champion in test_champions:
    for opponent in test_champions:

        if champion == opponent:
            continue

        try:
            result = matchup(
                champion1=champion,
                champion2=opponent,
                lane="top",
                rank="emerald"
            )

            # Convert JSON string -> Python dictionary
            if isinstance(result, str):
                result = json.loads(result)

            winrate = float(result["winrate"])

            games = int(
                result["number_of_games"].replace(",", "")
            )

            rows.append({
                "patch": "16.19",
                "rank": "emerald",
                "lane": "top",
                "champion": champion,
                "opponent": opponent,
                "expected_winrate": winrate,
                "matchup_games": games
            })

            print(
                f"{champion} vs {opponent}: "
                f"{winrate}% | {games} games"
            )

        except Exception as e:
            print(f"FAILED: {champion} vs {opponent}: {e}")

        time.sleep(0.2)


test_df = pd.DataFrame(rows)

print(test_df)

test_df.to_csv(
    "test_expected_winrates.csv",
    index=False
)
```
