```python
from lolalytics_api import matchup

result = matchup(
    champion1="riven",
    champion2="malphite",
    lane="top",
    rank="emerald"
)

print(result)
```

```python

tests = [
    ("riven", "malphite"),
    ("riven", "garen"),
    ("riven", "fiora"),
    ("riven", "camille"),
    ("riven", "irelia"),
]

for champ, opponent in tests:
    result = matchup(
        champion1=champ,
        champion2=opponent,
        lane="top",
        rank="emerald"
    )

    print(f"{champ.title()} vs {opponent.title()}: {result}")
```
